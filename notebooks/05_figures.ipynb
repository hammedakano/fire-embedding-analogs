```python
import sysfrom pathlib import Pathsys.path.insert(0, str(Path.cwd().parent / "src"))from fireanalogs.config import Configcfg = Config.load(Path.cwd().parent / "config" / "default.yaml")DATA = Path.cwd().parent / cfg["paths"]["data_root"]RESULTS = Path.cwd().parent / cfg["paths"]["results_root"]RESULTS.mkdir(parents=True, exist_ok=True)print("data:", DATA)
```

```python
import pandas as pdfrom fireanalogs import figures, validationloo = pd.read_csv(RESULTS / "loo_results.csv", parse_dates=["event_date"])metrics = pd.read_csv(RESULTS / "loo_metrics.csv", index_col=0)k_table = pd.read_csv(RESULTS / "k_sensitivity.csv")figures.loo_metrics_bars(metrics, RESULTS / "fig1_loo_metrics.png")figures.predicted_vs_observed(loo, RESULTS / "fig2_pred_vs_obs.png")relationship = validation.similarity_error_relationship(loo)figures.similarity_vs_error(loo, relationship["by_quartile"], RESULTS / "fig3_cosine_vs_error.png")figures.k_sensitivity(k_table, cfg.k_default, RESULTS / "fig4_k_sensitivity.png")print("event-level figures written")
```
