# Stage 5. Manuscript figures

Every figure is written from the tables produced upstream, so figures can be
regenerated without rerunning retrieval or validation.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent / "src"))

from fireanalogs.config import Config

cfg = Config.load(Path.cwd().parent / "config" / "default.yaml")
DATA = Path.cwd().parent / cfg["paths"]["data_root"]
RESULTS = Path.cwd().parent / cfg["paths"]["results_root"]
RESULTS.mkdir(parents=True, exist_ok=True)
print("data:", DATA)
```

```python
import pandas as pd
from fireanalogs import figures, validation

loo = pd.read_csv(RESULTS / "loo_results.csv", parse_dates=["event_date"])
metrics = pd.read_csv(RESULTS / "loo_metrics.csv", index_col=0)
k_table = pd.read_csv(RESULTS / "k_sensitivity.csv")

figures.loo_metrics_bars(metrics, RESULTS / "fig1_loo_metrics.png")
figures.predicted_vs_observed(loo, RESULTS / "fig2_pred_vs_obs.png")

relationship = validation.similarity_error_relationship(loo)
figures.similarity_vs_error(loo, relationship["by_quartile"], RESULTS / "fig3_cosine_vs_error.png")

figures.k_sensitivity(k_table, cfg.k_default, RESULTS / "fig4_k_sensitivity.png")
print("event-level figures written")
```

## Per-fire figures

```python
from fireanalogs import pixel

FIRE_PREFIX = "20260111_265_streatham_yalla_y_poora_road_name_fi"
rasters = pixel.load_fire(DATA / "rasters", FIRE_PREFIX, high_class_min=cfg.high_class_min)
summary = pixel.brier_summary(rasters)

figures.spatial_validation(rasters, summary, RESULTS / "fig_spatial_validation_RSE.png")

reliability_tables = {
    "Embedding": pixel.reliability(rasters, rasters.p_high_emb),
    "Geographic": pixel.reliability(rasters, rasters.p_high_geo),
}
figures.reliability_diagram(reliability_tables, RESULTS / "fig_reliability_diagram_RSE.png")

calibration = pd.read_csv(RESULTS / "calibration_fuelslope.csv")
figures.calibration_heatmap(calibration, RESULTS / "fig_calibration_fuelslope_RSE.png")
print("per-fire figures written")
```
